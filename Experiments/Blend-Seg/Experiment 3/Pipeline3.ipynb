```python
import os, io, cv2, numpy as np, pandas as pd, torch
from PIL import Image
import matplotlib.pyplot as plt
from tqdm import tqdm
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, random_split
from ultralytics import YOLO
import segmentation_models_pytorch as smp

############################################
# 1. Mappings & Data Loading
############################################
csv_path = "/content/drive/MyDrive/foodseg103/class_mappings.csv"
df_mapping = pd.read_csv(csv_path)

yolo_to_unet = {}
id_to_food = {0: "background"}
for _, row in df_mapping.iterrows():
    class_id = int(row["Class Id"])
    if class_id > 0:
        yolo_id = class_id - 1  # YOLO classes: 0-based
        unet_id = class_id      # U-Net classes: 0=background, 1..103=food
        yolo_to_unet[yolo_id] = unet_id
        id_to_food[unet_id] = row["Class Name"]

print("YOLO to U-Net mapping:", yolo_to_unet)
print("ID to Food mapping:", id_to_food)

train_paths = [
    "/content/drive/MyDrive/foodseg103/train-00000-of-00003-6bb37ec387d1825a.parquet",
    "/content/drive/MyDrive/foodseg103/train-00001-of-00003-4a1caa37147c0681.parquet",
    "/content/drive/MyDrive/foodseg103/train-00002-of-00003-c8b698399244cd95.parquet"
]
dfs = [pd.read_parquet(p) for p in train_paths]
full_df = pd.concat(dfs, ignore_index=True)
print(f"Combined training dataset contains {len(full_df)} samples.")

############################################
# 2. Helper: Extract Image & GT Mask
############################################
def extract_image_and_gt(row):
    img_bytes = row['image'].get("bytes")
    if img_bytes is None:
        raise ValueError("No image bytes found.")
    img = Image.open(io.BytesIO(img_bytes)).convert("RGB")

    mask_bytes = row['label'].get("bytes")
    if mask_bytes is None:
        raise ValueError("No mask bytes found.")
    gt_mask = Image.open(io.BytesIO(mask_bytes))
    return img, gt_mask

############################################
# 3. Load YOLO Model
############################################
# Updated YOLO model path
yolo_model_path = "/content/drive/MyDrive/food_yolo/results/yolov8x_detect_improved7/weights/best.pt"
yolo_model = YOLO(yolo_model_path)

def run_yolo_once(img):
    results = yolo_model(img, verbose=False)
    if len(results) == 0 or results[0].boxes is None:
        return []
    return results[0].boxes

############################################
# 4. Load Frozen U-Net++ (Updated Model)
############################################
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

# Updated U-Net++ model with EfficientNet-B4 encoder
unet_model = smp.UnetPlusPlus(
    encoder_name="efficientnet-b4",   # More powerful encoder
    encoder_weights="imagenet",
    in_channels=3,
    classes=104,
)
unet_model.to(DEVICE)

# Updated checkpoint path for U-Net++
checkpoint = torch.load("/content/drive/MyDrive/best_unetplusplus_checkpoint.pth", map_location=DEVICE)
unet_model.load_state_dict(checkpoint["model_state_dict"])
unet_model.eval()
print("U-Net++ model loaded on", DEVICE)

def unet_logits_full_image(img, target_size=(512,512)):
    img_resized = img.resize(target_size, Image.BILINEAR)
    img_arr = np.array(img_resized)
    img_tensor = torch.from_numpy(img_arr).permute(2,0,1).float().to(DEVICE) / 255.0
    mean = torch.tensor([0.485, 0.456, 0.406], device=DEVICE).view(3,1,1)
    std  = torch.tensor([0.229, 0.224, 0.225], device=DEVICE).view(3,1,1)
    img_tensor = (img_tensor - mean) / (std + 1e-8)
    img_tensor = img_tensor.unsqueeze(0)
    with torch.no_grad():
        logits = unet_model(img_tensor)
    return logits  # shape [1, 104, H, W]

############################################
# 5. Multi-Channel YOLO Map Generation with Scaling
############################################
def get_yolo_map(img, target_size=(512,512), pad=10, conf_thresh=0.3, alpha=5.0, num_classes=104):
    orig_w, orig_h = img.size
    H, W = target_size[1], target_size[0]
    # Create a multi-channel map: one channel per class (excluding background)
    yolo_map = np.zeros((num_classes, H, W), dtype=np.float32)
    boxes = run_yolo_once(img)
    if len(boxes) == 0:
        return yolo_map
    scale_x = W / orig_w
    scale_y = H / orig_h
    for box in boxes:
        coords = box.xyxy[0].cpu().numpy()  # [x1, y1, x2, y2]
        conf = float(box.conf.cpu().numpy()[0])
        if conf < conf_thresh:
            continue
        yolo_cls = int(box.cls.cpu().numpy()[0])
        if yolo_cls not in yolo_to_unet:
            continue
        unet_class = yolo_to_unet[yolo_cls]  # 1-indexed class
        channel = unet_class - 1  # convert to 0-index
        x1 = int(max(0, (coords[0] - pad) * scale_x))
        y1 = int(max(0, (coords[1] - pad) * scale_y))
        x2 = int(min(W, (coords[2] + pad) * scale_x))
        y2 = int(min(H, (coords[3] + pad) * scale_y))
        yolo_map[channel, y1:y2, x1:x2] = np.maximum(yolo_map[channel, y1:y2, x1:x2], alpha * conf)
    return yolo_map

############################################
# 6. BlendSeg: Residual Correction with Dual-Branch Fusion
############################################
class BlendSeg(nn.Module):
    def __init__(self, unet_channels, yolo_channels, num_classes, init_yolo_scale=5.0):
        super(BlendSeg, self).__init__()
        # Learnable scaling factor for YOLO map.
        self.yolo_scale = nn.Parameter(torch.tensor(init_yolo_scale, dtype=torch.float32))
        # Preprocessing branch for U-Net logits
        self.unet_branch = nn.Sequential(
            nn.Conv2d(unet_channels, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True)
        )
        # Preprocessing branch for YOLO map
        self.yolo_branch = nn.Sequential(
            nn.Conv2d(yolo_channels, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True)
        )
        # Fusion layers
        self.fusion = nn.Sequential(
            nn.Conv2d(64*2, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.Conv2d(64, num_classes, kernel_size=1)
        )

    def forward(self, unet_logits, yolo_map):
        yolo_scaled = self.yolo_scale * yolo_map
        unet_feat = self.unet_branch(unet_logits)
        yolo_feat = self.yolo_branch(yolo_scaled)
        fused = torch.cat([unet_feat, yolo_feat], dim=1)
        delta = self.fusion(fused)
        return unet_logits + delta  # residual correction

############################################
# 7. Simple Data Augmentation
############################################
def simple_augment(img: Image.Image, mask: Image.Image):
    if np.random.rand() < 0.5:
        img = img.transpose(Image.FLIP_LEFT_RIGHT)
        mask = mask.transpose(Image.FLIP_LEFT_RIGHT)
    return img, mask

############################################
# 8. Dataset Definition
############################################
class FoodSegDataset(Dataset):
    def __init__(self, df, target_size=(512,512), pad=10, conf_thresh=0.3, alpha=5.0, do_augment=False, num_classes=104):
        self.df = df
        self.target_size = target_size
        self.pad = pad
        self.conf_thresh = conf_thresh
        self.alpha = alpha
        self.do_augment = do_augment
        self.num_classes = num_classes

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        img, gt_mask_img = extract_image_and_gt(row)
        if self.do_augment:
            img, gt_mask_img = simple_augment(img, gt_mask_img)

        with torch.no_grad():
            unet_logits = unet_logits_full_image(img, target_size=self.target_size)
        unet_logits = unet_logits.squeeze(0)  # [104, H, W]

        yolo_map_np = get_yolo_map(img, target_size=self.target_size, pad=self.pad,
                                   conf_thresh=self.conf_thresh, alpha=self.alpha, num_classes=self.num_classes)
        yolo_map_tensor = torch.from_numpy(yolo_map_np)  # [104, H, W]

        gt_mask_resized = gt_mask_img.resize(self.target_size, Image.NEAREST)
        gt_mask_np = np.array(gt_mask_resized)
        gt_mask_tensor = torch.from_numpy(gt_mask_np).long()  # [H, W]

        return unet_logits, yolo_map_tensor, gt_mask_tensor, img

############################################
# 9. Custom Collate Function
############################################
def my_collate(batch):
    unet_logits = torch.stack([b[0] for b in batch])
    yolo_map = torch.stack([b[1] for b in batch])
    gt_mask = torch.stack([b[2] for b in batch])
    orig_imgs = [b[3] for b in batch]
    return unet_logits, yolo_map, gt_mask, orig_imgs

############################################
# 10. Train/Validation Split & DataLoaders
############################################
dataset = FoodSegDataset(full_df, target_size=(512,512), pad=10, conf_thresh=0.3, alpha=5.0, do_augment=True, num_classes=104)
val_fraction = 0.2
val_size = int(len(dataset) * val_fraction)
train_size = len(dataset) - val_size
train_dataset, val_dataset = random_split(dataset, [train_size, val_size])
print(f"Training samples: {len(train_dataset)}, Validation samples: {len(val_dataset)}")

batch_size = 6
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True, num_workers=0, collate_fn=my_collate)
val_loader   = DataLoader(val_dataset, batch_size=batch_size, shuffle=False, num_workers=0, collate_fn=my_collate)

############################################
# 11. Loss Functions: Focal Loss + Dice Loss
############################################
class FocalLoss(nn.Module):
    def __init__(self, gamma=2.0, alpha=None, reduction='mean'):
        super(FocalLoss, self).__init__()
        self.gamma = gamma
        self.alpha = alpha
        self.reduction = reduction

    def forward(self, inputs, targets):
        ce_loss = nn.functional.cross_entropy(inputs, targets, reduction='none')
        pt = torch.exp(-ce_loss)
        loss = ((1 - pt) ** self.gamma) * ce_loss
        if self.alpha is not None:
            if not torch.is_tensor(self.alpha):
                self.alpha = torch.tensor(self.alpha).to(inputs.device)
            at = self.alpha[targets]
            loss = at * loss
        if self.reduction == 'mean':
            return loss.mean()
        elif self.reduction == 'sum':
            return loss.sum()
        else:
            return loss

focal_loss_fn = FocalLoss(gamma=2.0, alpha=None, reduction='mean')

def dice_loss(outputs, targets, smooth=1e-6):
    outputs = torch.softmax(outputs, dim=1)
    targets_one_hot = torch.zeros_like(outputs)
    targets_one_hot.scatter_(1, targets.unsqueeze(1), 1)
    intersection = (outputs * targets_one_hot).sum(dim=(2,3))
    union = outputs.sum(dim=(2,3)) + targets_one_hot.sum(dim=(2,3))
    dice = (2 * intersection + smooth) / (union + smooth)
    return 1 - dice.mean()

def combined_loss_fn(outputs, targets):
    focal = focal_loss_fn(outputs, targets)
    dice  = dice_loss(outputs, targets)
    return focal + dice

############################################
# 12. Optimizer, Scheduler, and BlendSeg Setup
############################################
num_epochs = 60
blend_seg = BlendSeg(unet_channels=104, yolo_channels=104, num_classes=104, init_yolo_scale=5.0).to(DEVICE)
optimizer = optim.Adam(blend_seg.parameters(), lr=1e-4, weight_decay=1e-5)
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=2, verbose=True)

############################################
# 13. Segmentation Metrics
############################################
def compute_iou(pred, gt, num_classes=104):
    ious = []
    for cls in range(num_classes):
        pred_cls = (pred == cls).astype(np.uint8)
        gt_cls = (gt == cls).astype(np.uint8)
        intersection = np.sum(pred_cls & gt_cls)
        union = np.sum(pred_cls | gt_cls)
        if union == 0:
            continue
        ious.append(intersection / union)
    return np.mean(ious) if ious else 0.0

def compute_dice_metric(pred, gt, num_classes=104):
    dices = []
    for cls in range(num_classes):
        pred_cls = (pred == cls).astype(np.uint8)
        gt_cls = (gt == cls).astype(np.uint8)
        intersection = np.sum(pred_cls * gt_cls)
        total = np.sum(pred_cls) + np.sum(gt_cls)
        if total == 0:
            continue
        dices.append((2 * intersection) / total)
    return np.mean(dices) if dices else 0.0

############################################
# 14. Visualization Function
############################################
def visualize_samples(batch):
    blend_seg.eval()
    unet_logits, yolo_map, gt_mask, orig_imgs = batch
    unet_logits = unet_logits.to(DEVICE)
    yolo_map    = yolo_map.to(DEVICE)
    with torch.no_grad():
        outputs = blend_seg(unet_logits, yolo_map)
    for i in range(len(orig_imgs)):
        pred_mask = torch.argmax(outputs[i], dim=0).cpu().numpy()
        fig, axes = plt.subplots(1, 3, figsize=(15,5))
        axes[0].imshow(orig_imgs[i])
        axes[0].set_title("Original Image")
        axes[0].axis("off")
        axes[1].imshow(gt_mask[i].cpu().numpy(), cmap="jet")
        axes[1].set_title("Ground Truth")
        axes[1].axis("off")
        axes[2].imshow(pred_mask, cmap="jet")
        axes[2].set_title("BlendSeg Prediction")
        axes[2].axis("off")
        plt.show()

############################################
# 15. Training Loop with Validation & Visualization
############################################
best_val_loss = float('inf')
global_pix_acc = []
global_iou = []
global_dice = []

for epoch in range(num_epochs):
    blend_seg.train()
    running_loss = 0.0

    for unet_logits, yolo_map, gt_mask, orig_imgs in tqdm(train_loader, desc=f"Epoch {epoch+1}/{num_epochs} - Train"):
        unet_logits = unet_logits.to(DEVICE)
        yolo_map    = yolo_map.to(DEVICE)
        gt_mask     = gt_mask.to(DEVICE)

        optimizer.zero_grad()
        output = blend_seg(unet_logits, yolo_map)
        loss = combined_loss_fn(output, gt_mask)
        loss.backward()
        optimizer.step()
        running_loss += loss.item()

    avg_train_loss = running_loss / len(train_loader)

    blend_seg.eval()
    running_val_loss = 0.0
    epoch_iou = []
    epoch_dice = []
    epoch_pix_acc = []
    for unet_logits, yolo_map, gt_mask, orig_imgs in tqdm(val_loader, desc=f"Epoch {epoch+1}/{num_epochs} - Val"):
        unet_logits = unet_logits.to(DEVICE)
        yolo_map    = yolo_map.to(DEVICE)
        gt_mask     = gt_mask.to(DEVICE)
        with torch.no_grad():
            output = blend_seg(unet_logits, yolo_map)
            loss_val = combined_loss_fn(output, gt_mask)
            running_val_loss += loss_val.item()

            for i in range(output.shape[0]):
                pred_mask = torch.argmax(output[i], dim=0).cpu().numpy()
                gt_mask_np = gt_mask[i].cpu().numpy()
                pix_acc = np.sum(pred_mask == gt_mask_np) / gt_mask_np.size
                epoch_pix_acc.append(pix_acc)
                epoch_iou.append(compute_iou(pred_mask, gt_mask_np, 104))
                epoch_dice.append(compute_dice_metric(pred_mask, gt_mask_np, 104))

    avg_val_loss = running_val_loss / len(val_loader)
    mean_val_iou = np.mean(epoch_iou)
    mean_val_dice = np.mean(epoch_dice)
    mean_pix_acc = np.mean(epoch_pix_acc)

    global_pix_acc.append(mean_pix_acc)
    global_iou.append(mean_val_iou)
    global_dice.append(mean_val_dice)

    print(f"[Epoch {epoch+1}/{num_epochs}] "
          f"Train Loss: {avg_train_loss:.4f} | Val Loss: {avg_val_loss:.4f} "
          f"| Mean IoU: {mean_val_iou:.4f} | Mean Dice: {mean_val_dice:.4f} "
          f"| Pixel Acc: {mean_pix_acc:.4f} | YOLO Scale: {blend_seg.yolo_scale.item():.2f}")

    scheduler.step(avg_val_loss)

    if avg_val_loss < best_val_loss:
        best_val_loss = avg_val_loss
        torch.save(blend_seg.state_dict(), "/content/drive/MyDrive/foodseg103/blend_seg_best.pth")
        print(f"New best model saved (Val Loss: {best_val_loss:.4f})")

    if epoch + 1 == 3:
        print("Visualizing sample outputs after 3 epochs:")
        sample_batch = next(iter(val_loader))
        visualize_samples(sample_batch)

print("Training complete!")
print("\n======== Overall Validation Metrics ========")
print(f"Average Pixel Accuracy: {np.mean(global_pix_acc)*100:.2f}%")
print(f"Average Mean IoU:       {np.mean(global_iou)*100:.2f}%")
print(f"Average Mean Dice:      {np.mean(global_dice)*100:.2f}%")
print("==============================================")
```

```python
import os, io, cv2, numpy as np, pandas as pd, torch
from PIL import Image, ImageEnhance, ImageDraw, ImageFont
import matplotlib.pyplot as plt
from tqdm import tqdm
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import segmentation_models_pytorch as smp
from ultralytics import YOLO

#######################################################
# 1. Mappings & Data Loading
#######################################################
csv_path = "/content/drive/MyDrive/foodseg103/class_mappings.csv"
df_mapping = pd.read_csv(csv_path)
yolo_to_unet = {}
id_to_food = {0: "background"}
for _, row in df_mapping.iterrows():
    class_id = int(row["Class Id"])
    if class_id > 0:
        yolo_id = class_id - 1  # YOLO: 0-indexed
        unet_id = class_id      # U-Net: 0=background, 1..103=food
        yolo_to_unet[yolo_id] = unet_id
        id_to_food[unet_id] = row["Class Name"]

print("YOLO to U-Net mapping:", yolo_to_unet)
print("ID to Food mapping:", id_to_food)

#######################################################
# 2. Helper Functions & Dataset Definitions
#######################################################
def extract_image_and_gt(row):
    img_bytes = row['image'].get("bytes")
    if img_bytes is None:
        raise ValueError("No image bytes found.")
    img = Image.open(io.BytesIO(img_bytes)).convert("RGB")

    mask_bytes = row['label'].get("bytes")
    if mask_bytes is None:
        raise ValueError("No mask bytes found.")
    gt_mask = Image.open(io.BytesIO(mask_bytes))
    return img, gt_mask

# Simple augmentation (only used if do_augment=True)
def simple_augment(img: Image.Image, mask: Image.Image):
    if np.random.rand() < 0.5:
        img = img.transpose(Image.FLIP_LEFT_RIGHT)
        mask = mask.transpose(Image.FLIP_LEFT_RIGHT)
    angle = np.random.uniform(-10, 10)
    img = img.rotate(angle, resample=Image.BILINEAR)
    mask = mask.rotate(angle, resample=Image.NEAREST)
    return img, mask

class FoodSegDataset(Dataset):
    def __init__(self, df, target_size=(512,512), pad=10, conf_thresh=0.3,
                 alpha=5.0, do_augment=False, num_classes=104):
        self.df = df
        self.target_size = target_size
        self.pad = pad
        self.conf_thresh = conf_thresh
        self.alpha = alpha
        self.do_augment = do_augment
        self.num_classes = num_classes

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        img, gt_mask_img = extract_image_and_gt(row)
        if self.do_augment:
            img, gt_mask_img = simple_augment(img, gt_mask_img)

        # Get U-Net logits for the full image
        with torch.no_grad():
            unet_logits = unet_logits_full_image(img, target_size=self.target_size)
        unet_logits = unet_logits.squeeze(0)  # [104, H, W]

        # Generate multi-channel YOLO map
        yolo_map_np = get_yolo_map(img, target_size=self.target_size, pad=self.pad,
                                   conf_thresh=self.conf_thresh, alpha=self.alpha, num_classes=self.num_classes)
        yolo_map_tensor = torch.from_numpy(yolo_map_np)  # [104, H, W]

        # Resize GT mask to target size
        gt_mask_resized = gt_mask_img.resize(self.target_size, Image.NEAREST)
        gt_mask_np = np.array(gt_mask_resized)
        gt_mask_tensor = torch.from_numpy(gt_mask_np).long()  # [H, W]

        return unet_logits, yolo_map_tensor, gt_mask_tensor, img

def my_collate(batch):
    unet_logits = torch.stack([b[0] for b in batch])
    yolo_map = torch.stack([b[1] for b in batch])
    gt_mask = torch.stack([b[2] for b in batch])
    orig_imgs = [b[3] for b in batch]
    return unet_logits, yolo_map, gt_mask, orig_imgs

def compute_iou(pred, gt, num_classes=104):
    ious = []
    # Exclude background class (assumed to be class 0)
    for cls in range(1, num_classes):  # start from 1 instead of 0
        pred_cls = (pred == cls).astype(np.uint8)
        gt_cls = (gt == cls).astype(np.uint8)
        intersection = np.sum(pred_cls & gt_cls)
        union = np.sum(pred_cls | gt_cls)
        if union == 0:
            continue
        ious.append(intersection / union)
    return np.mean(ious) if ious else 0.0

def compute_dice_metric(pred, gt, num_classes=104):
    dices = []
    # Exclude background class (assumed to be class 0)
    for cls in range(1, num_classes):
        pred_cls = (pred == cls).astype(np.uint8)
        gt_cls = (gt == cls).astype(np.uint8)
        intersection = np.sum(pred_cls * gt_cls)
        total = np.sum(pred_cls) + np.sum(gt_cls)
        if total == 0:
            continue
        dices.append((2 * intersection) / total)
    return np.mean(dices) if dices else 0.0


#######################################################
# 3. Model Loading: U-Net++, YOLO & BlendSeg
#######################################################
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

# Load U-Net++ model with EfficientNet-B4 encoder (updated)
unet_model = smp.UnetPlusPlus(
    encoder_name="efficientnet-b4",   # More powerful encoder
    encoder_weights="imagenet",
    in_channels=3,
    classes=104,
)
unet_model.to(DEVICE)
checkpoint = torch.load("/content/drive/MyDrive/best_unetplusplus_checkpoint.pth", map_location=DEVICE)
unet_model.load_state_dict(checkpoint["model_state_dict"])
unet_model.eval()
print("U-Net++ model loaded on", DEVICE)

def unet_logits_full_image(img, target_size=(512,512)):
    img_resized = img.resize(target_size, Image.BILINEAR)
    img_arr = np.array(img_resized)
    img_tensor = torch.from_numpy(img_arr).permute(2,0,1).float().to(DEVICE) / 255.0
    mean = torch.tensor([0.485, 0.456, 0.406], device=DEVICE).view(3,1,1)
    std  = torch.tensor([0.229, 0.224, 0.225], device=DEVICE).view(3,1,1)
    img_tensor = (img_tensor - mean) / (std + 1e-8)
    img_tensor = img_tensor.unsqueeze(0)
    with torch.no_grad():
        logits = unet_model(img_tensor)
    return logits  # shape [1, 104, H, W]

# Load YOLO model (updated path)
yolo_model_path = "/content/drive/MyDrive/food_yolo/results/yolov8x_detect_improved7/weights/best.pt"
yolo_model = YOLO(yolo_model_path)

def run_yolo_once(img):
    results = yolo_model(img, verbose=False)
    if len(results) == 0 or results[0].boxes is None:
        return []
    return results[0].boxes

def get_yolo_map(img, target_size=(512,512), pad=10, conf_thresh=0.3, alpha=5.0, num_classes=104):
    orig_w, orig_h = img.size
    H, W = target_size[1], target_size[0]
    yolo_map = np.zeros((num_classes, H, W), dtype=np.float32)
    boxes = run_yolo_once(img)
    if len(boxes) == 0:
        return yolo_map
    scale_x = W / orig_w
    scale_y = H / orig_h
    for box in boxes:
        coords = box.xyxy[0].cpu().numpy()  # [x1, y1, x2, y2]
        conf = float(box.conf.cpu().numpy()[0])
        if conf < conf_thresh:
            continue
        yolo_cls = int(box.cls.cpu().numpy()[0])
        if yolo_cls not in yolo_to_unet:
            continue
        unet_class = yolo_to_unet[yolo_cls]
        channel = unet_class - 1  # convert to 0-index
        x1 = int(max(0, (coords[0] - pad) * scale_x))
        y1 = int(max(0, (coords[1] - pad) * scale_y))
        x2 = int(min(W, (coords[2] + pad) * scale_x))
        y2 = int(min(H, (coords[3] + pad) * scale_y))
        yolo_map[channel, y1:y2, x1:x2] = np.maximum(yolo_map[channel, y1:y2, x1:x2], alpha * conf)
    yolo_map = np.clip(yolo_map, 0, 10)
    return yolo_map

# Define BlendSeg architecture
class BlendSeg(nn.Module):
    def __init__(self, unet_channels, yolo_channels, num_classes, init_yolo_scale=5.0):
        super(BlendSeg, self).__init__()
        self.yolo_scale = nn.Parameter(torch.tensor(init_yolo_scale, dtype=torch.float32))
        self.unet_branch = nn.Sequential(
            nn.Conv2d(unet_channels, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True)
        )
        self.yolo_branch = nn.Sequential(
            nn.Conv2d(yolo_channels, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True)
        )
        # Removed dropout layer to match saved state
        self.fusion = nn.Sequential(
            nn.Conv2d(64*2, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.Conv2d(64, num_classes, kernel_size=1)
        )

    def forward(self, unet_logits, yolo_map):
        yolo_scaled = self.yolo_scale * yolo_map
        unet_feat = self.unet_branch(unet_logits)
        yolo_feat = self.yolo_branch(yolo_scaled)
        fused = torch.cat([unet_feat, yolo_feat], dim=1)
        delta = self.fusion(fused)
        return unet_logits + delta   # residual correction

blend_seg = BlendSeg(unet_channels=104, yolo_channels=104, num_classes=104, init_yolo_scale=5.0).to(DEVICE)
blend_seg.load_state_dict(torch.load("/content/drive/MyDrive/foodseg103/blend_seg_best.pth", map_location=DEVICE))
blend_seg.eval()

#######################################################
# 4. Visualization Helpers
#######################################################
def draw_yolo_boxes(img, boxes, pad=10):
    draw = ImageDraw.Draw(img)
    try:
        font = ImageFont.truetype("arial.ttf", 16)
    except IOError:
        font = ImageFont.load_default()
    for box in boxes:
        coords = box.xyxy[0].cpu().numpy()
        conf = float(box.conf.cpu().numpy()[0])
        if conf < 0.3:
            continue
        x1, y1, x2, y2 = coords
        x1, y1, x2, y2 = x1 - pad, y1 - pad, x2 + pad, y2 + pad
        draw.rectangle([x1, y1, x2, y2], outline="red", width=2)
        yolo_cls = int(box.cls.cpu().numpy()[0])
        if yolo_cls in yolo_to_unet:
            unet_cls = yolo_to_unet[yolo_cls]
            food_name = id_to_food.get(unet_cls, "unknown")
        else:
            food_name = "unknown"
        draw.text((x1, y1), f"{food_name} {conf:.2f}", fill="yellow", font=font)
    return img

def visualize_sample(sample_data, sample_metrics, title_suffix=""):
    orig_img = sample_data['orig_img']
    raw_pred = sample_data['raw_unet_pred']
    fusion_pred = sample_data['fusion_pred']
    gt_mask = sample_data['gt_mask']
    yolo_boxes = sample_data['yolo_boxes']

    # Overlay YOLO bounding boxes on a copy of the original image
    img_with_boxes = orig_img.copy()
    img_with_boxes = draw_yolo_boxes(img_with_boxes, yolo_boxes, pad=10)

    # Get food names (excluding background)
    def get_food_names(mask):
        cls_ids = set(np.unique(mask)) - {0}
        return [id_to_food.get(cls, str(cls)) for cls in cls_ids]
    fusion_foods = get_food_names(fusion_pred)
    gt_foods = get_food_names(gt_mask)

    # Plot 4 panels: Original+YOLO, Raw U-Net, Blend-Seg, Ground Truth with annotations.
    fig, axes = plt.subplots(1, 4, figsize=(20,5))
    axes[0].imshow(img_with_boxes)
    axes[0].set_title("Original Image + YOLO Boxes")
    axes[0].axis("off")

    axes[1].imshow(raw_pred, cmap="jet")
    axes[1].set_title("Raw U-Net Output")
    axes[1].axis("off")

    axes[2].imshow(fusion_pred, cmap="jet")
    axes[2].set_title("Blend-Seg Output")
    axes[2].axis("off")

    axes[3].imshow(gt_mask, cmap="jet")
    axes[3].set_title("Ground Truth")
    axes[3].axis("off")
    axes[3].text(0.5, -0.1,
                 f"Predicted Foods: {fusion_foods}\nGT Foods: {gt_foods}\nFood F1: {sample_metrics['food_f1']:.2f}",
                 transform=axes[3].transAxes, fontsize=10, ha="center", va="center")

    plt.suptitle(f"Sample Validation {title_suffix}\n"
                 f"IoU: {sample_metrics['iou']:.2f} | Dice: {sample_metrics['dice']:.2f} | Pixel Acc: {sample_metrics['pix_acc']:.2f}")
    plt.tight_layout(rect=[0, 0, 1, 0.90])
    plt.show()

def compute_food_f1(pred_mask, gt_mask):
    pred_foods = set(np.unique(pred_mask)) - {0}
    gt_foods   = set(np.unique(gt_mask)) - {0}
    if len(pred_foods) + len(gt_foods) == 0:
        return 1.0
    return 2 * len(pred_foods.intersection(gt_foods)) / (len(pred_foods) + len(gt_foods))

#######################################################
# 5. Main Validation Function
#######################################################
def validate_model():
    val_parquet = "/content/drive/MyDrive/foodseg103/validation-00000-of-00001-a5bfdaa5beb7006a.parquet"
    df_val = pd.read_parquet(val_parquet)
    print(f"Validation dataset contains {len(df_val)} samples.")

    # Create validation dataset & dataloader (no augmentation)
    val_dataset = FoodSegDataset(df_val, target_size=(512,512), pad=10, conf_thresh=0.3,
                                 alpha=5.0, do_augment=False, num_classes=104)
    val_loader = DataLoader(val_dataset, batch_size=1, shuffle=False, num_workers=0, collate_fn=my_collate)

    sample_results = []

    blend_seg.eval()
    with torch.no_grad():
        for idx, (unet_logits, yolo_map, gt_mask, orig_imgs) in enumerate(tqdm(val_loader, desc="Validating")):
            unet_logits = unet_logits.to(DEVICE)
            yolo_map = yolo_map.to(DEVICE)
            gt_mask = gt_mask.to(DEVICE)
            orig_img = orig_imgs[0]

            # Raw U-Net prediction via argmax
            raw_logits = unet_logits.squeeze(0)  # [104, H, W]
            raw_unet_pred = torch.argmax(raw_logits, dim=0).cpu().numpy()

            # BlendSeg output prediction
            fusion_out = blend_seg(unet_logits, yolo_map)  # [1, 104, H, W]
            fusion_pred = torch.argmax(fusion_out.squeeze(0), dim=0).cpu().numpy()

            gt_mask_np = gt_mask.squeeze(0).cpu().numpy()

            # Compute metrics
            iou_val = compute_iou(fusion_pred, gt_mask_np, num_classes=104)
            dice_val = compute_dice_metric(fusion_pred, gt_mask_np, num_classes=104)
            pix_acc = np.sum(fusion_pred == gt_mask_np) / gt_mask_np.size
            food_f1 = compute_food_f1(fusion_pred, gt_mask_np)

            # Run YOLO to get bounding boxes (for visualization)
            yolo_boxes = run_yolo_once(orig_img)

            sample_results.append({
                "index": idx,
                "orig_img": orig_img,
                "raw_unet_pred": raw_unet_pred,
                "fusion_pred": fusion_pred,
                "gt_mask": gt_mask_np,
                "iou": iou_val,
                "dice": dice_val,
                "pix_acc": pix_acc,
                "food_f1": food_f1,
                "yolo_boxes": yolo_boxes
            })

    # Overall averages
    avg_iou = np.mean([s["iou"] for s in sample_results])
    avg_dice = np.mean([s["dice"] for s in sample_results])
    avg_pix_acc = np.mean([s["pix_acc"] for s in sample_results])
    avg_food_f1 = np.mean([s["food_f1"] for s in sample_results])

    print("\n======== Overall Validation Metrics ========")
    print(f"Average Mean IoU:       {avg_iou*100:.2f}%")
    print(f"Average Mean Dice:      {avg_dice*100:.2f}%")
    print(f"Average Pixel Accuracy: {avg_pix_acc*100:.2f}%")
    print(f"Average Food F1 Score:  {avg_food_f1:.2f}")
    print("==============================================")

    # Visualize first 10 samples
    print("\nVisualizing first 25 validation samples...")
    for sample in sample_results[:25]:
        sample_metrics = {
            "iou": sample["iou"],
            "dice": sample["dice"],
            "pix_acc": sample["pix_acc"],
            "food_f1": sample["food_f1"]
        }
        visualize_sample(sample, sample_metrics, title_suffix=f"(Sample {sample['index']})")

    # Select best and worst samples based on IoU
    sorted_results = sorted(sample_results, key=lambda x: x["iou"], reverse=True)
    best_sample = sorted_results[2]
    worst_samples = sorted_results[-3:]

    print("\nVisualizing best performing sample (based on IoU):")
    best_metrics = {
        "iou": best_sample["iou"],
        "dice": best_sample["dice"],
        "pix_acc": best_sample["pix_acc"],
        "food_f1": best_sample["food_f1"]
    }
    visualize_sample(best_sample, best_metrics, title_suffix="(Best Performing)")

    print("\nVisualizing worst performing samples (based on IoU):")
    for sample in worst_samples:
        sample_metrics = {
            "iou": sample["iou"],
            "dice": sample["dice"],
            "pix_acc": sample["pix_acc"],
            "food_f1": sample["food_f1"]
        }
        visualize_sample(sample, sample_metrics, title_suffix=f"(Sample {sample['index']})")

    # Save detailed per-sample metrics to CSV for further comparison
    metrics_df = pd.DataFrame([{
        "index": s["index"],
        "iou": s["iou"],
        "dice": s["dice"],
        "pixel_accuracy": s["pix_acc"],
        "food_f1": s["food_f1"]
    } for s in sample_results])
    metrics_df.to_csv("/content/drive/MyDrive/foodseg103/Blend-Segvalidation_metrics.csv", index=False)
    print("Detailed per-sample metrics saved to validation_metrics.csv")

if __name__ == '__main__':
    validate_model()
```
