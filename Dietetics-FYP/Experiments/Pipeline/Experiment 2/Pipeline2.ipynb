```python
import os, io, cv2, numpy as np, pandas as pd, torch
from PIL import Image
import matplotlib.pyplot as plt
from tqdm import tqdm
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, random_split

############################################
# 1. Mappings & Data Loading
############################################
csv_path = "/content/drive/MyDrive/foodseg103/class_mappings.csv"
df_mapping = pd.read_csv(csv_path)

yolo_to_unet = {}
id_to_food = {0: "background"}
for _, row in df_mapping.iterrows():
    if int(row["Class Id"]) > 0:
        yolo_id = int(row["Class Id"]) - 1  # YOLO classes: 0-based
        unet_id = int(row["Class Id"])      # U-Net classes: 0=background, 1..103=food
        yolo_to_unet[yolo_id] = unet_id
        id_to_food[unet_id] = row["Class Name"]

print("YOLO to U-Net mapping:", yolo_to_unet)
print("ID to Food mapping:", id_to_food)

train_paths = [
    "/content/drive/MyDrive/foodseg103/train-00000-of-00003-6bb37ec387d1825a.parquet",
    "/content/drive/MyDrive/foodseg103/train-00001-of-00003-4a1caa37147c0681.parquet",
    "/content/drive/MyDrive/foodseg103/train-00002-of-00003-c8b698399244cd95.parquet"
]
dfs = [pd.read_parquet(p) for p in train_paths]
full_df = pd.concat(dfs, ignore_index=True)
print(f"Combined training dataset contains {len(full_df)} samples.")

############################################
# 2. Helper: Extract Image & GT Mask
############################################
def extract_image_and_gt(row):
    img_bytes = row['image'].get("bytes")
    if img_bytes is None:
        raise ValueError("No image bytes found.")
    img = Image.open(io.BytesIO(img_bytes)).convert("RGB")

    mask_bytes = row['label'].get("bytes")
    if mask_bytes is None:
        raise ValueError("No mask bytes found.")
    gt_mask = Image.open(io.BytesIO(mask_bytes))
    return img, gt_mask

############################################
# 3. Load YOLO Model
############################################
from ultralytics import YOLO
yolo_model_path = "/content/drive/MyDrive/food_yolo/results/yolov8l_detect_balanced/weights/best.pt"
yolo_model = YOLO(yolo_model_path)

def run_yolo_once(img):
    results = yolo_model(img, verbose=False)
    if len(results) == 0 or results[0].boxes is None:
        return []
    return results[0].boxes

############################################
# 4. Load Frozen U-Net
############################################
import segmentation_models_pytorch as smp
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
unet_model = smp.Unet(
    encoder_name="resnet50",
    encoder_weights="imagenet",
    in_channels=3,
    classes=104,
)
unet_model.to(DEVICE)
checkpoint = torch.load("/content/drive/MyDrive/foodseg103/unet_checkpoint_model2.pth", map_location=DEVICE)
unet_model.load_state_dict(checkpoint["model_state_dict"])
unet_model.eval()
print("U-Net model loaded on", DEVICE)

def unet_logits_full_image(img, target_size=(512,512)):
    img_resized = img.resize(target_size, Image.BILINEAR)
    img_arr = np.array(img_resized)
    img_tensor = torch.from_numpy(img_arr).permute(2,0,1).float().to(DEVICE) / 255.0
    mean = torch.tensor([0.485, 0.456, 0.406], device=DEVICE).view(3,1,1)
    std  = torch.tensor([0.229, 0.224, 0.225], device=DEVICE).view(3,1,1)
    img_tensor = (img_tensor - mean) / (std + 1e-8)
    img_tensor = img_tensor.unsqueeze(0)
    with torch.no_grad():
        logits = unet_model(img_tensor)
    return logits  # shape [1, 104, H, W]

############################################
# 5. YOLO Map Generation w/ Scaling
############################################
def get_yolo_map(img, target_size=(512,512), pad=10, conf_thresh=0.3, alpha=5.0):
    orig_w, orig_h = img.size
    yolo_map = np.zeros((target_size[1], target_size[0]), dtype=np.float32)
    boxes = run_yolo_once(img)
    for box in boxes:
        coords = box.xyxy[0].cpu().numpy()
        conf = float(box.conf.cpu().numpy()[0])
        if conf < conf_thresh:
            continue
        scale_x = target_size[0] / orig_w
        scale_y = target_size[1] / orig_h
        x1, y1, x2, y2 = coords
        x1 = int(max(0, (x1 - pad) * scale_x))
        y1 = int(max(0, (y1 - pad) * scale_y))
        x2 = int(min(target_size[0], (x2 + pad) * scale_x))
        y2 = int(min(target_size[1], (y2 + pad) * scale_y))
        yolo_map[y1:y2, x1:x2] = np.maximum(yolo_map[y1:y2, x1:x2], alpha * conf)
    return yolo_map

############################################
# 6. FusionNet: Residual Correction with Learnable YOLO Scaling
############################################
class FusionNet(nn.Module):
    def __init__(self, unet_channels, yolo_channels, num_classes, init_yolo_scale=5.0):
        super(FusionNet, self).__init__()
        # Learnable scaling factor for YOLO map.
        self.yolo_scale = nn.Parameter(torch.tensor(init_yolo_scale, dtype=torch.float32))
        in_channels = unet_channels + yolo_channels  # typically 104 + 1 = 105
        self.conv1 = nn.Conv2d(in_channels, 64, kernel_size=3, padding=1)
        self.bn1   = nn.BatchNorm2d(64)
        self.relu  = nn.ReLU(inplace=True)
        self.conv2 = nn.Conv2d(64, 64, kernel_size=3, padding=1)
        self.bn2   = nn.BatchNorm2d(64)
        self.conv3 = nn.Conv2d(64, num_classes, kernel_size=1)

    def forward(self, unet_logits, yolo_map):
        yolo_scaled = self.yolo_scale * yolo_map
        x = torch.cat([unet_logits, yolo_scaled], dim=1)  # [B, 105, H, W]
        x = self.relu(self.bn1(self.conv1(x)))
        x = self.relu(self.bn2(self.conv2(x)))
        delta = self.conv3(x)
        return unet_logits + delta  # residual correction

############################################
# 7. (Removed) Extra Coverage Loss
############################################
# We removed the extra bounding-box coverage loss.

############################################
# 8. Simple Data Augmentation
############################################
def simple_augment(img: Image.Image, mask: Image.Image):
    if np.random.rand() < 0.5:
        img = img.transpose(Image.FLIP_LEFT_RIGHT)
        mask = mask.transpose(Image.FLIP_LEFT_RIGHT)
    return img, mask

############################################
# 9. Dataset Definition
############################################
class FoodSegDataset(Dataset):
    def __init__(self, df, target_size=(512,512), pad=10, conf_thresh=0.3, alpha=5.0, do_augment=False):
        self.df = df
        self.target_size = target_size
        self.pad = pad
        self.conf_thresh = conf_thresh
        self.alpha = alpha
        self.do_augment = do_augment

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        img, gt_mask_img = extract_image_and_gt(row)
        if self.do_augment:
            img, gt_mask_img = simple_augment(img, gt_mask_img)

        with torch.no_grad():
            unet_logits = unet_logits_full_image(img, target_size=self.target_size)
        unet_logits = unet_logits.squeeze(0)  # [104, H, W]

        yolo_map_np = get_yolo_map(img, target_size=self.target_size, pad=self.pad,
                                   conf_thresh=self.conf_thresh, alpha=self.alpha)
        yolo_map_tensor = torch.from_numpy(yolo_map_np).unsqueeze(0)  # [1, H, W]

        gt_mask_resized = gt_mask_img.resize(self.target_size, Image.NEAREST)
        gt_mask_np = np.array(gt_mask_resized)
        gt_mask_tensor = torch.from_numpy(gt_mask_np).long()  # [H, W]

        return unet_logits, yolo_map_tensor, gt_mask_tensor, img

############################################
# 10. Custom Collate Function
############################################
def my_collate(batch):
    unet_logits = torch.stack([b[0] for b in batch])
    yolo_map = torch.stack([b[1] for b in batch])
    gt_mask = torch.stack([b[2] for b in batch])
    orig_imgs = [b[3] for b in batch]
    return unet_logits, yolo_map, gt_mask, orig_imgs

############################################
# 11. Train/Validation Split & DataLoaders
############################################
dataset = FoodSegDataset(full_df, target_size=(512,512), pad=10, conf_thresh=0.3, alpha=5.0, do_augment=True)
val_fraction = 0.2
val_size = int(len(dataset) * val_fraction)
train_size = len(dataset) - val_size
train_dataset, val_dataset = random_split(dataset, [train_size, val_size])
print(f"Training samples: {len(train_dataset)}, Validation samples: {len(val_dataset)}")

batch_size = 4
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True, num_workers=0, collate_fn=my_collate)
val_loader   = DataLoader(val_dataset, batch_size=batch_size, shuffle=False, num_workers=0, collate_fn=my_collate)

############################################
# 12. Loss Functions: Focal Loss + Dice Loss
############################################
class FocalLoss(nn.Module):
    def __init__(self, gamma=2.0, alpha=None, reduction='mean'):
        super(FocalLoss, self).__init__()
        self.gamma = gamma
        self.alpha = alpha  # scalar or tensor of class weights
        self.reduction = reduction

    def forward(self, inputs, targets):
        ce_loss = nn.functional.cross_entropy(inputs, targets, reduction='none')
        pt = torch.exp(-ce_loss)
        loss = ((1 - pt) ** self.gamma) * ce_loss
        if self.alpha is not None:
            if not torch.is_tensor(self.alpha):
                self.alpha = torch.tensor(self.alpha).to(inputs.device)
            at = self.alpha[targets]
            loss = at * loss
        if self.reduction == 'mean':
            return loss.mean()
        elif self.reduction == 'sum':
            return loss.sum()
        else:
            return loss

focal_loss_fn = FocalLoss(gamma=2.0, alpha=None, reduction='mean')

def dice_loss(outputs, targets, smooth=1e-6):
    outputs = torch.softmax(outputs, dim=1)
    targets_one_hot = torch.zeros_like(outputs)
    targets_one_hot.scatter_(1, targets.unsqueeze(1), 1)
    intersection = (outputs * targets_one_hot).sum(dim=(2,3))
    union = outputs.sum(dim=(2,3)) + targets_one_hot.sum(dim=(2,3))
    dice = (2 * intersection + smooth) / (union + smooth)
    return 1 - dice.mean()

def combined_loss_fn(outputs, targets):
    focal = focal_loss_fn(outputs, targets)
    dice  = dice_loss(outputs, targets)
    return focal + dice

############################################
# 13. Optimizer, Scheduler, and FusionNet Setup
############################################
num_epochs = 30
fusion_net = FusionNet(unet_channels=104, yolo_channels=1, num_classes=104, init_yolo_scale=5.0).to(DEVICE)
optimizer = optim.Adam(fusion_net.parameters(), lr=1e-4, weight_decay=1e-5)
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=2, verbose=True)

############################################
# 14. Segmentation Metrics
############################################
def compute_iou(pred, gt, num_classes=104):
    ious = []
    for cls in range(num_classes):
        pred_cls = (pred == cls).astype(np.uint8)
        gt_cls = (gt == cls).astype(np.uint8)
        intersection = np.sum(pred_cls & gt_cls)
        union = np.sum(pred_cls | gt_cls)
        if union == 0:
            continue
        ious.append(intersection / union)
    return np.mean(ious) if ious else 0.0

def compute_dice_metric(pred, gt, num_classes=104):
    dices = []
    for cls in range(num_classes):
        pred_cls = (pred == cls).astype(np.uint8)
        gt_cls = (gt == cls).astype(np.uint8)
        intersection = np.sum(pred_cls * gt_cls)
        total = np.sum(pred_cls) + np.sum(gt_cls)
        if total == 0:
            continue
        dices.append((2 * intersection) / total)
    return np.mean(dices) if dices else 0.0

############################################
# 15. Visualization Function
############################################
def visualize_samples(batch):
    fusion_net.eval()
    unet_logits, yolo_map, gt_mask, orig_imgs = batch
    unet_logits = unet_logits.to(DEVICE)
    yolo_map    = yolo_map.to(DEVICE)
    with torch.no_grad():
        outputs = fusion_net(unet_logits, yolo_map)
    for i in range(len(orig_imgs)):
        pred_mask = torch.argmax(outputs[i], dim=0).cpu().numpy()
        fig, axes = plt.subplots(1, 3, figsize=(15,5))
        axes[0].imshow(orig_imgs[i])
        axes[0].set_title("Original Image")
        axes[0].axis("off")
        axes[1].imshow(gt_mask[i].cpu().numpy(), cmap="jet")
        axes[1].set_title("Ground Truth")
        axes[1].axis("off")
        axes[2].imshow(pred_mask, cmap="jet")
        axes[2].set_title("FusionNet Prediction")
        axes[2].axis("off")
        plt.show()

############################################
# 16. Training Loop with Validation & Visualization
############################################
best_val_loss = float('inf')

all_pix_acc = []  # to store pixel accuracies
all_miou = []
all_dice = []
all_food_f1 = []

for epoch in range(num_epochs):
    fusion_net.train()
    running_loss = 0.0

    for unet_logits, yolo_map, gt_mask, orig_imgs in tqdm(train_loader, desc=f"Epoch {epoch+1}/{num_epochs} - Train"):
        unet_logits = unet_logits.to(DEVICE)
        yolo_map    = yolo_map.to(DEVICE)
        gt_mask     = gt_mask.to(DEVICE)

        optimizer.zero_grad()
        output = fusion_net(unet_logits, yolo_map)
        loss = combined_loss_fn(output, gt_mask)
        loss.backward()
        optimizer.step()
        running_loss += loss.item()

    avg_train_loss = running_loss / len(train_loader)

    fusion_net.eval()
    running_val_loss = 0.0
    val_iou_list = []
    val_dice_list = []
    for unet_logits, yolo_map, gt_mask, orig_imgs in tqdm(val_loader, desc=f"Epoch {epoch+1}/{num_epochs} - Val"):
        unet_logits = unet_logits.to(DEVICE)
        yolo_map    = yolo_map.to(DEVICE)
        gt_mask     = gt_mask.to(DEVICE)
        with torch.no_grad():
            output = fusion_net(unet_logits, yolo_map)
            loss_val = combined_loss_fn(output, gt_mask)
            running_val_loss += loss_val.item()

            for i in range(output.shape[0]):
                pred_mask = torch.argmax(output[i], dim=0).cpu().numpy()
                gt_mask_np = gt_mask[i].cpu().numpy()
                # Compute pixel accuracy
                pix_acc = np.sum(pred_mask == gt_mask_np) / gt_mask_np.size
                val_iou_list.append(compute_iou(pred_mask, gt_mask_np, 104))
                val_dice_list.append(compute_dice_metric(pred_mask, gt_mask_np, 104))
                all_pix_acc.append(pix_acc)

    avg_val_loss = running_val_loss / len(val_loader)
    mean_val_iou = np.mean(val_iou_list)
    mean_val_dice = np.mean(val_dice_list)

    print(f"[Epoch {epoch+1}/{num_epochs}] "
          f"Train Loss: {avg_train_loss:.4f} | Val Loss: {avg_val_loss:.4f} "
          f"| Mean IoU: {mean_val_iou:.4f} | Mean Dice: {mean_val_dice:.4f} "
          f"| YOLO Scale: {fusion_net.yolo_scale.item():.2f}")

    scheduler.step(avg_val_loss)

    if avg_val_loss < best_val_loss:
        best_val_loss = avg_val_loss
        torch.save(fusion_net.state_dict(), "/content/drive/MyDrive/foodseg103/fusion_net_best.pth")
        print(f"New best model saved (Val Loss: {best_val_loss:.4f})")

    if epoch + 1 == 3:
        print("Visualizing sample outputs after 3 epochs:")
        sample_batch = next(iter(val_loader))
        visualize_samples(sample_batch)

print("Training complete!")

# Print overall validation metrics computed during training
print("\n======== Overall Validation Metrics ========")
print(f"Average Pixel Accuracy: {np.mean(all_pix_acc)*100:.2f}%")
print(f"Average Mean IoU:       {np.mean(all_miou)*100:.2f}%")
print(f"Average Mean Dice:      {np.mean(all_dice)*100:.2f}%")
print(f"Average Food F1:        {np.mean(all_food_f1)*100:.2f}%")
print("==============================================")
```

```python
import os, io, cv2, numpy as np, pandas as pd, torch
from PIL import Image
import matplotlib.pyplot as plt
from tqdm import tqdm
import segmentation_models_pytorch as smp
from ultralytics import YOLO
import torch.nn as nn

############################################
# 1. YOLO -> U-Net Mappings
############################################
csv_path = "/content/drive/MyDrive/foodseg103/class_mappings.csv"
df_mapping = pd.read_csv(csv_path)

yolo_to_unet = {}
id_to_food = {0: "background"}
for _, row in df_mapping.iterrows():
    if int(row["Class Id"]) > 0:
        yolo_id = int(row["Class Id"]) - 1  # YOLO classes are 0-based
        unet_id = int(row["Class Id"])      # U-Net: 0=background, 1..103=food
        yolo_to_unet[yolo_id] = unet_id
        id_to_food[unet_id] = row["Class Name"]

print("YOLO to U-Net mapping:", yolo_to_unet)
print("ID to Food mapping:", id_to_food)

############################################
# 2. Load Validation Parquet
############################################
val_parquet_path = "/content/drive/MyDrive/foodseg103/validation-00000-of-00001-a5bfdaa5beb7006a.parquet"
val_df = pd.read_parquet(val_parquet_path)
print(f"Loaded {len(val_df)} validation samples.")

############################################
# 3. Helper: Extract Image and Ground Truth
############################################
def extract_image_and_gt(row):
    img_bytes = row['image'].get("bytes")
    if img_bytes is None:
        raise ValueError("No image bytes found.")
    img = Image.open(io.BytesIO(img_bytes)).convert("RGB")

    mask_bytes = row['label'].get("bytes")
    if mask_bytes is None:
        raise ValueError("No mask bytes found.")
    gt_mask = Image.open(io.BytesIO(mask_bytes))
    return img, gt_mask

############################################
# 4. Load YOLO Model
############################################
yolo_model_path = "/content/drive/MyDrive/food_yolo/results/yolov8l_detect_balanced/weights/best.pt"
yolo_model = YOLO(yolo_model_path)
print("YOLO model loaded.")

def run_yolo_once(img):
    results = yolo_model(img, verbose=False)
    if len(results) == 0 or results[0].boxes is None:
        return []
    return results[0].boxes

############################################
# 5. Load Frozen U-Net Model
############################################
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
unet_model = smp.Unet(
    encoder_name="resnet50",
    encoder_weights="imagenet",
    in_channels=3,
    classes=104,
)
unet_model.to(DEVICE)
checkpoint = torch.load("/content/drive/MyDrive/foodseg103/unet_checkpoint_model2.pth", map_location=DEVICE)
unet_model.load_state_dict(checkpoint["model_state_dict"])
unet_model.eval()
print("U-Net model loaded on", DEVICE)

def unet_logits_full_image(img, target_size=(512,512)):
    img_resized = img.resize(target_size, Image.BILINEAR)
    img_arr = np.array(img_resized)
    img_tensor = torch.from_numpy(img_arr).permute(2,0,1).float().to(DEVICE) / 255.0
    mean = torch.tensor([0.485, 0.456, 0.406], device=DEVICE).view(3,1,1)
    std  = torch.tensor([0.229, 0.224, 0.225], device=DEVICE).view(3,1,1)
    img_tensor = (img_tensor - mean) / (std + 1e-8)
    img_tensor = img_tensor.unsqueeze(0)  # shape [1, 3, H, W]
    with torch.no_grad():
        logits = unet_model(img_tensor)
    return logits  # shape [1, 104, H, W]

############################################
# 6. YOLO Map Generation
############################################
def get_yolo_map(img, target_size=(512,512), pad=10, conf_thresh=0.3):
    orig_w, orig_h = img.size
    yolo_map = np.zeros((target_size[1], target_size[0]), dtype=np.float32)
    boxes = run_yolo_once(img)
    for box in boxes:
        coords = box.xyxy[0].cpu().numpy()
        conf = float(box.conf.cpu().numpy()[0])
        if conf < conf_thresh:
            continue
        scale_x = target_size[0] / orig_w
        scale_y = target_size[1] / orig_h
        x1, y1, x2, y2 = coords
        x1 = int(max(0, (x1 - pad) * scale_x))
        y1 = int(max(0, (y1 - pad) * scale_y))
        x2 = int(min(target_size[0], (x2 + pad) * scale_x))
        y2 = int(min(target_size[1], (y2 + pad) * scale_y))
        # Here we use the raw confidence; the learnable scaling in FusionNet will amplify it.
        yolo_map[y1:y2, x1:x2] = np.maximum(yolo_map[y1:y2, x1:x2], conf)
    return yolo_map  # shape (H, W)

############################################
# 7. Define FusionNet Model (Matching Training)
############################################
# This definition should match the one used during training.
class FusionNet(nn.Module):
    def __init__(self, unet_channels, yolo_channels, num_classes, init_yolo_scale=5.0):
        super(FusionNet, self).__init__()
        # Learnable scaling factor for YOLO map.
        self.yolo_scale = nn.Parameter(torch.tensor(init_yolo_scale, dtype=torch.float32))
        in_channels = unet_channels + yolo_channels  # e.g., 104 + 1 = 105
        self.conv1 = nn.Conv2d(in_channels, 64, kernel_size=3, padding=1)
        self.bn1   = nn.BatchNorm2d(64)
        self.relu  = nn.ReLU(inplace=True)
        self.conv2 = nn.Conv2d(64, 64, kernel_size=3, padding=1)
        self.bn2   = nn.BatchNorm2d(64)
        self.conv3 = nn.Conv2d(64, num_classes, kernel_size=1)

    def forward(self, unet_logits, yolo_map):
        # Multiply YOLO map by the learnable scaling factor.
        yolo_scaled = self.yolo_scale * yolo_map
        x = torch.cat([unet_logits, yolo_scaled], dim=1)  # [B, 105, H, W]
        x = self.relu(self.bn1(self.conv1(x)))
        x = self.relu(self.bn2(self.conv2(x)))
        delta = self.conv3(x)
        return unet_logits + delta  # residual correction

# Load the trained FusionNet checkpoint.
fusion_net = FusionNet(unet_channels=104, yolo_channels=1, num_classes=104, init_yolo_scale=5.0).to(DEVICE)
fusion_checkpoint = torch.load("/content/drive/MyDrive/foodseg103/fusion_net_best.pth", map_location=DEVICE)
fusion_net.load_state_dict(fusion_checkpoint)
fusion_net.eval()
print("FusionNet model loaded.")

############################################
# 8. Metric Functions
############################################
def compute_pixel_accuracy(pred, gt):
    return np.sum(pred == gt) / pred.size

def compute_iou(pred, gt, num_classes=104):
    ious = []
    for cls in range(num_classes):
        pred_cls = (pred == cls).astype(np.uint8)
        gt_cls = (gt == cls).astype(np.uint8)
        intersection = np.sum(pred_cls & gt_cls)
        union = np.sum(pred_cls | gt_cls)
        if union == 0:
            continue
        ious.append(intersection / union)
    return np.mean(ious) if ious else 0.0

def compute_dice(pred, gt, num_classes=104):
    dices = []
    for cls in range(num_classes):
        pred_cls = (pred == cls).astype(np.uint8)
        gt_cls = (gt == cls).astype(np.uint8)
        intersection = np.sum(pred_cls * gt_cls)
        total = np.sum(pred_cls) + np.sum(gt_cls)
        if total == 0:
            continue
        dices.append((2 * intersection) / total)
    return np.mean(dices) if dices else 0.0

def list_food_items(seg_mask):
    unique = np.unique(seg_mask)
    items = [id_to_food.get(cls, f"Class {cls}") for cls in unique if cls != 0]
    return items

def compute_food_f1(pred_items, true_items):
    pred_set = set(pred_items)
    true_set = set(true_items)
    if len(pred_set) == 0 and len(true_set) == 0:
        return 1.0
    intersection = len(pred_set.intersection(true_set))
    if len(pred_set) == 0:
        precision = 1.0 if intersection == 0 else 0.0
    else:
        precision = intersection / len(pred_set)
    if len(true_set) == 0:
        recall = 1.0 if intersection == 0 else 0.0
    else:
        recall = intersection / len(true_set)
    if precision + recall == 0:
        return 0.0
    return 2.0 * (precision * recall) / (precision + recall)

############################################
# 9. Validation Loop
############################################
all_pix_acc = []
all_iou = []
all_dice = []
all_food_f1 = []
MAX_DISPLAY = 25

for idx in range(len(val_df)):
    try:
        row = val_df.iloc[idx]
        img, gt_mask_img = extract_image_and_gt(row)
        # Resize GT mask to (512,512)
        gt_mask = np.array(gt_mask_img.resize((512,512), Image.NEAREST).convert("L"))

        # (A) Get frozen U-Net logits.
        full_img_logits = unet_logits_full_image(img, target_size=(512,512))  # [1, 104, H, W]

        # (B) Get YOLO map.
        yolo_map_np = get_yolo_map(img, target_size=(512,512), pad=10, conf_thresh=0.3)
        # Convert to tensor with shape [1, 1, H, W]
        yolo_map_tensor = torch.from_numpy(yolo_map_np).unsqueeze(0).unsqueeze(0).to(DEVICE)

        # (C) Get FusionNet output.
        with torch.no_grad():
            fusion_output = fusion_net(full_img_logits, yolo_map_tensor)  # [1, 104, H, W]
            fused_pred = torch.argmax(fusion_output, dim=1).squeeze(0).cpu().numpy()

        # (D) Get raw U-Net prediction for reference.
        raw_unet_pred = torch.argmax(full_img_logits, dim=1).squeeze(0).cpu().numpy()

        # (E) Compute Metrics.
        pix_acc = compute_pixel_accuracy(fused_pred, gt_mask)
        iou = compute_iou(fused_pred, gt_mask, 104)
        dice = compute_dice(fused_pred, gt_mask, 104)
        all_pix_acc.append(pix_acc)
        all_iou.append(iou)
        all_dice.append(dice)
        pred_items = list_food_items(fused_pred)
        true_items = list_food_items(gt_mask)
        food_f1 = compute_food_f1(pred_items, true_items)
        all_food_f1.append(food_f1)

        # (F) Visualization for the first MAX_DISPLAY samples.
        if idx < MAX_DISPLAY:
            fig, axes = plt.subplots(2, 2, figsize=(15,15))
            axes[0,0].imshow(img)
            axes[0,0].set_title("Original Image")
            axes[0,0].axis("off")

            # Draw YOLO boxes for visualization.
            yolo_box_img = np.array(img).copy()
            boxes = run_yolo_once(img)
            for box in boxes:
                coords = box.xyxy[0].cpu().numpy()
                conf_ = float(box.conf.cpu().numpy()[0])
                if conf_ < 0.3:
                    continue
                x1, y1, x2, y2 = map(int, coords)
                pad_ = 10
                x1 = max(0, x1 - pad_)
                y1 = max(0, y1 - pad_)
                x2 = min(img.width, x2 + pad_)
                y2 = min(img.height, y2 + pad_)
                cv2.rectangle(yolo_box_img, (x1,y1), (x2,y2), (0,255,0), 2)
            axes[0,1].imshow(yolo_box_img)
            axes[0,1].set_title("YOLO Boxes")
            axes[0,1].axis("off")

            axes[1,0].imshow(raw_unet_pred, cmap="jet")
            axes[1,0].set_title("Raw U-Net Prediction")
            axes[1,0].axis("off")

            axes[1,1].imshow(fused_pred, cmap="jet")
            axes[1,1].set_title("FusionNet Prediction")
            axes[1,1].axis("off")

            title_str = (
                f"Image {idx} Metrics:\n"
                f"Pixel Acc: {pix_acc:.4f}, IoU: {iou:.4f}, Dice: {dice:.4f}, FoodF1: {food_f1:.4f}\n"
                f"Predicted: {pred_items} | True: {true_items}"
            )
            plt.suptitle(title_str, fontsize=16)
            plt.tight_layout()
            plt.show()
    except Exception as e:
        print(f"Error processing image {idx}: {e}")

avg_pix_acc = np.mean(all_pix_acc)
avg_iou = np.mean(all_iou)
avg_dice = np.mean(all_dice)
avg_food_f1 = np.mean(all_food_f1)

print("\n======== Overall Validation Metrics ========")
print(f"Average Pixel Accuracy: {avg_pix_acc*100:.2f}%")
print(f"Average Mean IoU:       {avg_iou*100:.2f}%")
print(f"Average Mean Dice:      {avg_dice*100:.2f}%")
print(f"Average Food F1:        {avg_food_f1*100:.2f}%")
print("==============================================")
```
